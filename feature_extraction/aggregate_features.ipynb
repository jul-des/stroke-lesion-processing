### Feature aggregation
This notebook provides one way to aggregate the per lesion features from the extract_features notebook to per patient features. We used this feature aggregation in out paper. 

First, you can specify for which features the patient mean should be taken for aggregation, we used this for total lesion volume. Then you can spefify for which features the patient mean and standard deviation should be taken for aggregation. We did this for the shape and and per lesion volume features. Finally, the aggregate_lesion_features function will automatically calcuate different patient level metrics of lesion distribution, including distances between lesions in MNI space, the number of affected territories, whether lesions are bilateral and many more. A full list with detailed definitions can be found in the supplemental material of our paper. 

```python
data = pd.read_csv("path/to/per/lesion/data.csv")
```

```python
col_mean = ['Total Lesion Volume']

col_mean_sd = ['Individual lesion volume','MeshVolume', 'SurfaceArea', 'SurfaceVolumeRatio', 'Sphericity', 'Maximum3DDiameter', 'Maximum2DDiameterSlice', 
               'Maximum2DDiameterColumn', 'Maximum2DDiameterRow', 'MajorAxisLength', 'MinorAxisLength', 'LeastAxisLength', 'Elongation', 'Flatness']
```

```python
agg_data = aggregate_lesion_features(data, mean_columns=col_mean, mean_sd_columns=col_mean_sd)
```

```python
agg_data.to_csv("path/to/save.csv")
```
